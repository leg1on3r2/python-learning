# Домашняя работа 1. Класс Chocolate

```python
class Chocolate:
    def __init__(self, price: float, fill: str = ""):
        if price <= 0:
            raise ValueError("Цена батончика должна быть больше нуля")
        self.price = price
        self.fill = fill

    def show_my_chocolate(self):
        if self.fill:
            print(f"Шоколадный батончик, добавка: {self.fill}")
        else:
            print("Обычный шоколадный батончик")

    def __str__(self):
        return f"Шоколадный батончик, {self.price} рублей"
```

```python
bar_with_fill = Chocolate(55.5, "орехи")
bar_without_fill = Chocolate(42.0)

bar_with_fill.show_my_chocolate()
bar_without_fill.show_my_chocolate()

print("-" * 30)

print(bar_with_fill)
print(bar_without_fill)
```

```python
for bad_price in (-10, 0):
    try:
        Chocolate(bad_price)
    except ValueError as error:
        print(f"Цена {bad_price}: ValueError — {error}")
```
